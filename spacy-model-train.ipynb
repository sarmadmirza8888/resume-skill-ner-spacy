```python
def find_overlaps(annotations):
    overlaps = []

    # Sort by start position, then end position
    sorted_annotations = sorted(
        annotations,
        key=lambda x: (x[0], x[1])
    )

    for i in range(len(sorted_annotations)):
        start1, end1, label1 = sorted_annotations[i]

        for j in range(i + 1, len(sorted_annotations)):
            start2, end2, label2 = sorted_annotations[j]

            # If the second entity starts before the first one ends
            if start2 < end1:
                overlaps.append(
                    (sorted_annotations[i], sorted_annotations[j])
                )

            # Since sorted by start, no later entity can overlap
            # once its start is >= end1
            else:
                break

    return overlaps


total_overlapping_files = 0
total_overlapping_pairs = 0

for data in all_data:
    overlaps = find_overlaps(data["annotations"])

    if overlaps:
        total_overlapping_files += 1
        total_overlapping_pairs += len(overlaps)

print("Files containing overlapping annotations:", total_overlapping_files)
print("Total overlapping annotation pairs:", total_overlapping_pairs)
```

```python
def remove_overlaps_keep_longest(annotations):
    # Sort by:
    # 1. start position
    # 2. longer span first
    sorted_annotations = sorted(
        annotations,
        key=lambda x: (x[0], -(x[1] - x[0]))
    )

    kept = []

    for annotation in sorted_annotations:

        start, end, label = annotation

        # Check whether this annotation overlaps
        # with something we already kept
        overlaps = False

        for kept_annotation in kept:

            kept_start, kept_end, kept_label = kept_annotation

            if start < kept_end and end > kept_start:
                overlaps = True
                break

        # Keep only non-overlapping annotations
        if not overlaps:
            kept.append(annotation)

    return kept


# ------------------------------------------------------------
# Test on all 500 files
# ------------------------------------------------------------

total_before = 0
total_after = 0
files_changed = 0

for data in all_data:

    annotations = data["annotations"]

    cleaned_annotations = remove_overlaps_keep_longest(
        annotations
    )

    total_before += len(annotations)
    total_after += len(cleaned_annotations)

    if len(cleaned_annotations) != len(annotations):
        files_changed += 1


print("Total annotations BEFORE cleaning:", total_before)
print("Total annotations AFTER cleaning :", total_after)
print("Files changed                    :", files_changed)
print("Annotations removed              :", total_before - total_after)
```

```python
import spacy

# Create a blank English pipeline
nlp = spacy.blank("en")

valid_entities = 0
invalid_entities = 0
misaligned_entities = 0

invalid_examples = []

for data in normalized_data:

    text = data["text"]
    doc = nlp.make_doc(text)

    for start, end, label in data["annotations"]:

        # Check that character positions are valid
        if start < 0 or end > len(text) or start >= end:
            invalid_entities += 1

            if len(invalid_examples) < 10:
                invalid_examples.append(
                    (start, end, text[start:end])
                )

            continue

        # Check whether spaCy can align the span
        span = doc.char_span(
            start,
            end,
            label=label,
            alignment_mode="strict"
        )

        if span is None:
            misaligned_entities += 1

            if len(invalid_examples) < 10:
                invalid_examples.append(
                    (start, end, text[start:end])
                )
        else:
            valid_entities += 1


print("Valid entities       :", valid_entities)
print("Invalid entities     :", invalid_entities)
print("Misaligned entities  :", misaligned_entities)

print("\nTotal checked        :", 
      valid_entities + invalid_entities + misaligned_entities)

if invalid_examples:
    print("\nSome problematic examples:")

    for start, end, entity_text in invalid_examples:
        print(
            "Start:", start,
            "| End:", end,
            "| Text:", repr(entity_text)
        )
```

```python
# Step 5.1 — Compare strict vs expanded alignment

strict_count = 0
expanded_count = 0
still_failed = 0

examples = []

for data in normalized_data:

    text = data["text"]
    doc = nlp.make_doc(text)

    for start, end, label in data["annotations"]:

        # Strict alignment
        strict_span = doc.char_span(
            start,
            end,
            label=label,
            alignment_mode="strict"
        )

        if strict_span is not None:
            strict_count += 1
            continue

        # Try expanding the annotation to token boundaries
        expanded_span = doc.char_span(
            start,
            end,
            label=label,
            alignment_mode="expand"
        )

        if expanded_span is not None:
            expanded_count += 1

            if len(examples) < 20:
                examples.append({
                    "original": text[start:end],
                    "expanded": expanded_span.text
                })
        else:
            still_failed += 1


print("Strictly aligned :", strict_count)
print("Expanded         :", expanded_count)
print("Still failed     :", still_failed)

print("\nExamples of expansion:")

for example in examples:
    print(
        repr(example["original"]),
        "→",
        repr(example["expanded"])
    )
```

```python
import spacy

nlp = spacy.blank("en")

training_data = []

removed_annotations = 0

for data in normalized_data:

    text = data["text"]
    doc = nlp.make_doc(text)

    valid_entities = []

    for start, end, label in data["annotations"]:

        span = doc.char_span(
            start,
            end,
            label=label,
            alignment_mode="strict"
        )

        if span is not None:
            valid_entities.append(
                (start, end, label)
            )
        else:
            removed_annotations += 1

    training_data.append(
        (
            text,
            {
                "entities": valid_entities
            }
        )
    )

print("Number of CVs:", len(training_data))

print(
    "Total usable annotations:",
    sum(
        len(item[1]["entities"])
        for item in training_data
    )
)

print(
    "Misaligned annotations skipped:",
    removed_annotations
)
```

```python
from sklearn.model_selection import train_test_split

train_data, test_data = train_test_split(
    training_data,
    test_size=0.20,
    random_state=42
)

print("Total CVs :", len(training_data))
print("Training CVs :", len(train_data))
print("Testing CVs  :", len(test_data))
```

```python
from spacy.training import Example

train_examples = []

for text, annotations in train_data:

    doc = nlp.make_doc(text)

    example = Example.from_dict(
        doc,
        annotations
    )

    train_examples.append(example)

print("Training examples:", len(train_examples))
```

```python
import spacy

# 1. Create blank English model
nlp = spacy.blank("en")

# 2. Add NER pipeline
ner = nlp.add_pipe("ner")

# 3. Add our entity label
ner.add_label("SKILL")

# 4. Initialize the model
optimizer = nlp.initialize(
    get_examples=lambda: train_examples
)

print("Pipeline:", nlp.pipe_names)
print("Labels:", ner.labels)
print("Model initialized successfully!")
```

```python
import random
from spacy.util import minibatch, compounding

# Number of times the model sees the training data
n_epochs = 10

for epoch in range(n_epochs):

    # Shuffle training examples
    random.shuffle(train_examples)

    losses = {}

    # Create mini-batches
    batches = minibatch(
        train_examples,
        size=compounding(4.0, 32.0, 1.5)
    )

    # Train on each batch
    for batch in batches:

        nlp.update(
            batch,
            drop=0.3,
            sgd=optimizer,
            losses=losses
        )

    print(
        f"Epoch {epoch + 1}/{n_epochs} "
        f"Loss: {losses.get('ner', 0):.4f}"
    )
```

```python
predictions = []

for text, annotations in test_data:

    doc = trained_nlp(text)

    predicted_entities = []

    for ent in doc.ents:
        predicted_entities.append(
            (
                ent.start_char,
                ent.end_char,
                ent.label_
            )
        )

    predictions.append(
        {
            "text": text,
            "actual": annotations["entities"],
            "predicted": predicted_entities
        }
    )

print("Test CVs evaluated:", len(predictions))
```

```python
TP = 0
FP = 0
FN = 0

for result in predictions:

    actual = set(
        result["actual"]
    )

    predicted = set(
        result["predicted"]
    )

    # Correct predictions
    TP += len(actual & predicted)

    # Predicted but not actually present
    FP += len(predicted - actual)

    # Actually present but not predicted
    FN += len(actual - predicted)


print("True Positives :", TP)
print("False Positives:", FP)
print("False Negatives:", FN)
```

```python
precision = (TP / (TP + FP))*100
recall = (TP / (TP + FN))*100

f1 = (2 * (precision * recall) / (precision + recall))*100

print("Precision :", precision)
print("Recall    :", recall)
print("F1 Score  :", f1)
```

```python
from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

y_true = []
y_pred = []

for result in predictions:

    text = result["text"]

    # Create spaCy document
    doc = trained_nlp.make_doc(text)

    actual_entities = result["actual"]
    predicted_entities = result["predicted"]

    for token in doc:

        token_start = token.idx
        token_end = token.idx + len(token.text)

        # -------------------------
        # ACTUAL LABEL
        # -------------------------

        actual_skill = False

        for start, end, label in actual_entities:

            if token_start >= start and token_end <= end:
                actual_skill = True
                break

        # -------------------------
        # PREDICTED LABEL
        # -------------------------

        predicted_skill = False

        for start, end, label in predicted_entities:

            if token_start >= start and token_end <= end:
                predicted_skill = True
                break

        # -------------------------
        # STORE LABELS
        # -------------------------

        if actual_skill:
            y_true.append("SKILL")
        else:
            y_true.append("NON-SKILL")

        if predicted_skill:
            y_pred.append("SKILL")
        else:
            y_pred.append("NON-SKILL")


print("Total tokens evaluated:", len(y_true))
```

```python
cm = confusion_matrix(
    y_true,
    y_pred,
    labels=["SKILL", "NON-SKILL"]
)

TP = cm[0][0]
FN = cm[0][1]
FP = cm[1][0]
TN = cm[1][1]

print("True Positives :", TP)
print("True Negatives :", TN)
print("False Positives:", FP)
print("False Negatives:", FN)
```

```python
TP_clean = 0
FP_clean = 0
FN_clean = 0

for result in clean_predictions:

    actual = set(result["actual"])
    predicted = set(result["predicted"])

    TP_clean += len(actual & predicted)
    FP_clean += len(predicted - actual)
    FN_clean += len(actual - predicted)

print("Clean Model Results:")
print("True Positives :", TP_clean)
print("False Positives:", FP_clean)
print("False Negatives:", FN_clean)
```

```python
precision_clean = TP_clean / (TP_clean + FP_clean)

recall_clean = TP_clean / (TP_clean + FN_clean)

f1_clean = (
    2 * precision_clean * recall_clean
    / (precision_clean + recall_clean)
)

print("Clean Model Metrics:")
print(f"Precision: {precision_clean * 100:.2f}%")
print(f"Recall   : {recall_clean * 100:.2f}%")
print(f"F1 Score : {f1_clean * 100:.2f}%")
```
